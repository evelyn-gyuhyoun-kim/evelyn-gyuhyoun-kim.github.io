```python
import pandas as pd
import matplotlib.pyplot as plt
import wikipediaapi as wk
%matplotlib inline

df = pd.read_csv("sports.csv")
df=df.set_index("SPORT")

operation=""
while (operation!="X"):
    if operation=="A":
        #DISPLAY ALL
        print("DISPLAY ALL")
        print(df)
        graph=input("Would you like a graph? (Y/N)")
        if graph.upper()=="Y":
            print(df.plot(kind="bar"))
            plt.show()
            
    elif operation=="S":
        #DISPLAY SELECTED RECORDS
        print("DISPLAY SELECTED RECORDS")
        sports=input("Please enter sports separated by a comma")
        sports=sports.split(",")
        countries=input("Please enter countries separated by a comma")
        countries=countries.split(",")
        print(df.loc[sports,countries])
        graph=input("Would you like a graph? (Y/N)")
        if graph.upper()=="Y":
            df.loc[sports,countries].plot(kind="bar")
            plt.show() 

    elif operation=="N":
        #ADD A RECORD
        print("ADD A SPORT")
        sport = input("Please enter sport: ")
        sport=sport.upper()
        scores=[]
        for i in range(len(df.columns)):
            score=int(input("PLEASE ENTER THE SCORE FOR "+df.columns[i]+": "))
            scores.append(score)

        df.loc[sport]=scores 
        
    elif operation=="D":
        #DELETE A RECORD
        print("DELETE A SPORT")
        sport = input("Please enter sport to delete: ")
        sport = sport.upper() 
        df = df.drop(sport)

    elif operation=="I":
        #DISPLAY FROM WIKI
        print("GO TO WIKIPEDIA")
        country = input("PLEASE ENTER A COUNTRY OR SPORT: ")

        wiki = wk.Wikipedia('CIS2300 (gyuhyoun.kim@example.com)', 'en')
        page = wiki.page(country)

        if page.exists():
            print("\n*** DATA FROM WIKIPEDIA ***")
            print(page.summary)
        else:
            print("Page not found on Wikipedia.")

    elif operation=="F":
        #SAVE THE DATA
        print("WRITE FILE")
        df.to_csv("sports_new.csv") 
        
    operation=input("\n***Main MENU***\n\nPLEASE SELECT AN OPERATION:\nA - Display All\nS - Display Selected Records\nN - New Record\nD - Delete Record\nI - Get Wikipedia\nF - Save the Data\nX - Exit System\nPlease Select Option: ").upper()

print("THANK YOU. HAVE A NICE DAY!")
```
